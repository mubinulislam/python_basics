# **Implementing Search Functionality**

```python
def search_mobile_issues(products, keyword):
    # List to store search results
    results = []

    # Convert the keyword to lowercase for case-insensitive search
    keyword_lower = keyword.lower()

    # Iterate over each product in the list
    for product in products:
        # Convert the 'Known_Issues' to lowercase
        issues_lower = product['Known_Issues'].lower()

        # Check if the keyword is in the 'Known_Issues'
        if keyword_lower in issues_lower:
            # If found, add the product to the results list
            results.append(product)

    # Return the list of matching products
    return results
```

# **Main Program Loop**

```python
def main():
    print("Welcome to the Mobile Issues Search Program!")

    while True:
        # Prompt the user to enter a keyword
        keyword = input("\nEnter an issue keyword to search for (or type 'exit' to quit): ")

        # Check if the user wants to exit
        if keyword.lower() == 'exit':
            print("Thank you for using the program. Goodbye!")
            break
        if not keyword:
            print("You didn't enter any keyword. Please try again.")
            continue

        # Search for mobile phones with matching issues
        matching_products = search_mobile_issues(mobile_products, keyword)

        # Check if any results were found
        if matching_products:
            print("\nMobile phones with issues matching your search:")
            # Display each matching product
            for product in matching_products:
                print(f"\nModel Name: {product['Model_Name']}")
                print(f"Features: {product['Features']}")
                print(f"Release Date: {product['Release_Date']}")
                print(f"Price: ${product['Price']}")
                print(f"Known Issues: {product['Known_Issues']}")
        else:
            print("\nNo mobile phones found with issues matching your search.")

if __name__ == "__main__":
    main()
```
